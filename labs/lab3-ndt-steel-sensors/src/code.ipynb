```python
from google.colab import drive
drive.mount('/drive')

path = '/drive/MyDrive/Magnetic Materials/'

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
compressive = pd.read_csv(path + 'compressive.csv', sep=';', )
tensile = pd.read_csv(path + 'tensile.csv', sep=';')

x_cmp=compressive.iloc[:, 0]
y_cmp=compressive.iloc[:, 1]

x_tns=tensile.iloc[:, 0]
y_tns=tensile.iloc[:, 1]

plt.figure(figsize=(8, 5))
plt.plot(x_cmp, y_cmp, label='Compressive')
plt.plot(x_tns, y_tns, label='Tensile')

plt.xlabel('σ (MPa)')
plt.ylabel('Β (mT)')
plt.title('Μαγνητική Επαγωγή Β συναρτήσει της τάσης σ')
plt.grid(True, linestyle='--', alpha=0.5)

plt.legend()
plt.tight_layout()
plt.show()
```

```python
data1 = pd.read_csv(path + 'Hall_sample1.csv', sep=';')
data2 = pd.read_csv(path + 'Hall_sample2.csv', sep=';')
data3 = pd.read_csv(path + 'Hall_sample3.csv', sep=';')

x1=data1.iloc[:, 0]
y1=data1.iloc[:, 1]

x2=data2.iloc[:, 0]
y2=data2.iloc[:, 1]

x3=data3.iloc[:, 0]
y3=data3.iloc[:, 1]

# Define the window size for the moving average
window_size = 5 # You can change this value

# Apply moving average to each sample's y-data
y1_smoothed = y1.rolling(window=window_size, min_periods=1).mean()
y2_smoothed = y2.rolling(window=window_size, min_periods=1).mean()
y3_smoothed = y3.rolling(window=window_size, min_periods=1).mean()

plt.figure(figsize=(10, 6))

# Plot original and smoothed data for Sample 1
plt.plot(x1, y1_smoothed, color='blue', label='Sample 1', alpha=0.7)

# Plot original and smoothed data for Sample 2
plt.plot(x2, y2_smoothed, color='red', label='Sample 2', alpha=0.7)

# Plot original and smoothed data for Sample 3
plt.plot(x3, y3_smoothed, color='orange', label='Sample 3', alpha=0.7)

plt.xlabel('Time (ms)')
plt.ylabel('Output Voltage Vout (mV)')
plt.title('Smoothed Hall Sensor Output Voltage (Vout) vs. Time (t)')
plt.grid(True, linestyle='--', alpha=0.5)

plt.legend()
plt.tight_layout()
plt.show()
```

```python
welded = pd.read_csv(path + 'welded.csv', sep=';')

time=welded.iloc[:, 0]
bx=welded.iloc[:, 1]
by=welded.iloc[:, 2]
bz=welded.iloc[:, 3]
b = np.sqrt(bx**2 + by**2 + bz**2)

plt.figure(figsize=(8, 5))
plt.plot(time, b, label='Magnetic Induction (B)')

plt.xlabel('Time (ms)')
plt.ylabel('Magnetic Induction Β (μT)')
plt.title('Magnetic Induction (B) vs. Time (t)')
plt.grid(True, linestyle='--', alpha=0.5)

plt.legend()
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

files = ['AMR_sample1.csv', 'AMR_sample2.csv', 'AMR_sample3.csv']
titles = [
    'Magnetic Induction (B) vs. Time for Sample 1',
    'Magnetic Induction (B) vs. Time for Sample 2',
    'Magnetic Induction (B) vs. Time for Sample 3'
]
colors = ['blue', 'red', 'green']

for file, title, color in zip(files, titles, colors):

    data = pd.read_csv(path + file, sep=';')

    time = data.iloc[:, 0]
    bx = data.iloc[:, 1]
    by = data.iloc[:, 2]
    bz = data.iloc[:, 3]

    b = np.sqrt(bx**2 + by**2 + bz**2)

    plt.figure(figsize=(6, 2))
    plt.plot(time, b, color=color)

    plt.title(title)
    plt.xlabel('Time (ms)')
    plt.ylabel('B (μT)')
    plt.grid(True, linestyle='--', alpha=0.5)

    plt.tight_layout()
    plt.show()
```
