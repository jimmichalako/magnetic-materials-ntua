```python
from google.colab import drive
drive.mount('/drive')

path = '/drive/MyDrive/Magnetic Materials/'

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Switch between 100mm.csv, 150mm.csv, 200mm.csv, 250mm.csv
data = pd.read_csv(path + '/250mm.csv')

time = data.iloc[:, 0]   # first column
vout = data.iloc[:, 1]   # second column

plt.figure(figsize=(8, 5))
plt.plot(time, vout, label='Distance between coils: 250mm') # Also switch this

plt.xlabel('time(s)')
plt.ylabel('vout(V)')
plt.title('')
plt.grid(True, linestyle='--', alpha=0.5)

plt.legend()
plt.tight_layout()
plt.show()
```

```python
x = np.array([0.100, 0.150, 0.200, 0.250])
y = [0.00002065, 0.00003185, 0.00004175, 0.00005225]

# Fit a line (y = m*x + b)
m, b = np.polyfit(x, y, 1)
y_fit = m * x + b

plt.figure(figsize=(7, 5))

# Plot scatter points
plt.scatter(x, y, color='blue', label='Data points')

# Plot fitted line
plt.plot(x, y_fit, color='red', label=f'Fitted line')

plt.xlabel("Distance (m)")
plt.ylabel("Time (s)")
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()

plt.show()
```

```python
# Switch between 5g.csv, 20g.csv, 100g.csv, 200g.csv
data = pd.read_csv(path + '/200g.csv')

time = data.iloc[:, 0]   # first column
vout = data.iloc[:, 1]   # second column

plt.figure(figsize=(8, 5))
plt.plot(time, vout, label='Mass: 200g') # Also switch this

plt.xlabel('time(s)')
plt.ylabel('vout(V)')
plt.title('')
plt.grid(True, linestyle='--', alpha=0.5)

plt.legend()
plt.tight_layout()
plt.show()
```

```python
x = np.array([0.0088, 0.02813, 0.04164, 0.04336])
y = [5, 20, 100, 200]

# Fit a quadratic curve (y = a*x^2 + b*x + c) to the data
a, b, c = np.polyfit(x, y, 2)

# Generate a continuous range of x values for plotting the smooth curve
x_curve = np.linspace(x.min(), x.max(), 100) # 100 points for a smooth curve
y_curve = a * x_curve**2 + b * x_curve + c

plt.figure(figsize=(7, 5))

# Plot scatter points
plt.scatter(x, y, color='blue', label='Data points')

# Plot the optimally fitted curve using the continuous x_curve
plt.plot(x_curve, y_curve, color='red', label=f'Fitted curve')

plt.xlabel("Vpp (V)")
plt.ylabel("Mass (g)")
plt.grid(True, linestyle='--', alpha=0.5)
plt.legend()
plt.tight_layout()

plt.show()
```
